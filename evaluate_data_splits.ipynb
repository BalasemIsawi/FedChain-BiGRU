```python
SEEDs = [42, 100, 1337]
IIDs = [True, False]
num_clients_list = [10, 15, 20]

test_list = []
test_attack_list = []
test_normal_list = []
test_attack_ratio_list = []

for num_clients in num_clients_list:
    for IID in IIDs:
        train_list = []
        train_attack_list = []
        train_normal_list = []
        train_attack_ratio_list = []

        for SEED in SEEDs:
            os.environ["PYTHONHASHSEED"] = str(SEED)
            random.seed(SEED)
            np.random.seed(SEED)

            X_train_clients, y_train, X_test, y_test = prepare_data(r"ton-iot-train-test-network/train_test_network.csv", 
                num_clients=num_clients, 
                iid=IID, 
                alpha=0.5, 
                test_size=0.2,
                SEED=SEED
            )

            train = [n.shape[0] for n in y_train]
            train_attack = [len(np.where(y == 1)[0]) for y in y_train]
            train_normal = [len(np.where(y == 0)[0]) for y in y_train]
            train_attack_ratio = [c/n for c, n in zip(train_attack, train)]

            train_list.append(train)
            train_attack_list.append(train_attack)
            train_normal_list.append(train_normal)
            train_attack_ratio_list.append(train_attack_ratio)

            if num_clients == num_clients_list[0] and IID == IIDs[0] and SEED == SEEDs[0]:
                test_attack = len(np.where(y_test == 1)[0])
                test_normal = len(np.where(y_test == 0)[0])
                test_attack_ratio = test_attack / len(y_test)

                test_list.append(len(y_test))
                test_attack_list.append(test_attack)
                test_normal_list.append(test_normal)
                test_attack_ratio_list.append(test_attack_ratio)

        
        print(f"K={num_clients}, IID={IID}")
        print(f"Train Windows Count: Min={int(np.min(train_list))}, Max={int(np.max(train_list))}, Mean={int(np.mean(train_list))} \u00B1{np.std(train_list)}")
        print(f"Train Attack Windows Count: Min={int(np.min(train_attack_list))}, Max={int(np.max(train_attack_list))}, Mean={int(np.mean(train_attack_list))} \u00B1{int(np.std(train_attack_list))}")
        print(f"Train Normal Windows Count: Min={int(np.min(train_normal_list))}, Max={int(np.max(train_normal_list))}, Mean={int(np.mean(train_normal_list))} \u00B1{int(np.std(train_normal_list))}")
        print(f"Train Windows Attack Ratio: Min={np.min(train_attack_ratio_list)*100:0.2f}%, Max={np.max(train_attack_ratio_list)*100:0.2f}%, Mean={np.mean(train_attack_ratio_list)*100:0.2f}% \u00B1{np.std(train_attack_ratio_list)*100:0.2f}%")
        print()
    print()
print()
print(f"Test Windows Count: Min={int(np.min(test_list))}, Max={int(np.max(test_list))}, Mean={int(np.mean(test_list))} \u00B1{np.std(test_list)}")
print(f"Test Attack Windows Count: Min={int(np.min(test_attack_list))}, Max={int(np.max(test_attack_list))}, Mean={int(np.mean(test_attack_list))} \u00B1{int(np.std(test_attack_list))}")
print(f"Test Normal Windows Count: Min={int(np.min(test_normal_list))}, Max={int(np.max(test_normal_list))}, Mean={int(np.mean(test_normal_list))} \u00B1{int(np.std(test_normal_list))}")
print(f"Test Windows Attack Ratio: Min={np.min(test_attack_ratio_list)*100:0.2f}%, Max={np.max(test_attack_ratio_list)*100:0.2f}%, Mean={np.mean(test_attack_ratio_list)*100:0.2f}% \u00B1{np.std(test_attack_ratio_list)*100:0.2f}%")

```
